```python
import pandas as pd
#ejemplo 1 Series de ventas de 5 productos
ventas = pd.Series(
    [250, 320, 150, 410, 90],
    index=["Producto A", "Producto B", "Producto C", "Producto D", "Producto E"]
)

ventas
```

```python
#Total y promedio de ventas
total_ventas = ventas.sum()
promedio_ventas = ventas.mean()

print("Total de ventas:", total_ventas)
print("Promedio de ventas:", promedio_ventas)
```

```python
#Filtrar productos con ventas mayores a 300 pesos
ventas_mayores_300 = ventas[ventas > 300]
ventas_mayores_300
```

```python
#ejemplo 2 Series con calificaciones de alumnos

calificaciones = pd.Series(
    [8.5, 9.2, 7.8, 6.9, 9.8],
    index=["Ana", "Luis", "María", "Pedro", "Sofía"]
)

calificaciones
```

```python
#Filtrar alumnos con calificación mayor a 9

calificaciones_superiores = calificaciones[calificaciones > 9]
calificaciones_superiores
```

```python
#ejemplo 3 Series con inventario de tienda

inventario = pd.Series(
    [12, 5, 0, 17, 9],
    index=["Playeras", "Pantalones", "Sudaderas", "Gorras", "Zapatos"]
)

inventario
```

```python
#Verificar qué productos están agotados

agotados = inventario[inventario == 0]
agotados
```

```python
#ejemplo 4 Series con precios de productos

precios = pd.Series(
    [19.99, 49.99, 15.50, 120.00, 75.50],
    index=["Camisa", "Pantalón", "Calcetas", "Chamarras", "Tenis"]
)

precios

```

```python
#Descuento del 10% aplicado a cada producto

precios_descuento = precios * 0.90
precios_descuento
```

```python
#ejemplo 5 Series de temperatura por ciudad

temperaturas = pd.Series(
    [22, 18, 25, 30, 27],
    index=["CDMX", "Toluca", "Guadalajara", "Monterrey", "Cancún"]
)

temperaturas
```

```python
#Filtrar ciudades con temperatura mayor a 25°C

calor = temperaturas[temperaturas > 25]
calor
```
