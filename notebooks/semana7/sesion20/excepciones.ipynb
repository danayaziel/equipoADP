#### 1. Conversión de tipos y validación numérica. 

```python
# Se ingresó 7.5 como prueba para el programa

try:
    valor = input("Ingresa un número entero: ")
    numero = int(valor)  
    print("El número es:", numero)

except ValueError:
    print("Error: Debes ingresar solo números enteros.")

else:
    print("No hubo errores durante la conversión.")

finally:
    print("Fin del programa")
```

#### 2. Lectura de archivos con control de errores

```python
archivo = "datos.txt"
try:
    with open(archivo, "r", encoding="utf-8") as f:
        contenido = f.read()
        print("Archivo leído correctamente:")
        print(contenido)

except FileNotFoundError:
    print(f"Archivo '{archivo}' no encontrado. Verifica el nombre e intenta de nuevo.")

finally:
    print("Fin del programa")
```

#### 3. Excepción personalizada lanzada con raise

```python
# Se ingresó -4 como prueba para el programa 

def verificar_precio(precio):
    """ Lanza una excepción si la edad es negativa """
    if precio <= 0:
        raise ValueError("El precio no puede ser negativo")
    return f"Precio registrado correctamente: ${precio:.2f}"

try:
    entrada = float(input("Ingresa el precio del producto: "))
    mensaje = verificar_precio(entrada)
    print(mensaje)

except ValueError as e:
    print("Error:", e)

finally:
    print("Fin del programa")
```

#### Extra: incluye un ejemplo con else y finally

```python
# Se usó 0 como prueba para el programa

try:
    numero = float(input("Ingresa un número para dividirlo entre 100: "))
    resultado = 100 / numero
except ZeroDivisionError:
    print("No puedes dividir entre cero")
except ValueError:
    print("Debes ingresar un número")
else:
    print("Resultado:", round(resultado,2))
finally:
    print("Fin del programa")
```
