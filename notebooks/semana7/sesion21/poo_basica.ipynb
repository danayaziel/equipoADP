```python
class Coche:
    def __init__(self, marca, modelo, anio):
        self.marca = marca
        self.modelo = modelo
        self.anio = anio

    def arrancar(self):
        return f"El {self.marca} {self.modelo} del {self.anio} está disponible en México"

# Ejemplos
c1 = Coche("Toyota", "Corolla", 2020)
print(c1.arrancar())

c2 = Coche("Honda", "Civic", 2018)
print(c2.arrancar()) 
```

```python
class Libro:
    def __init__(self, titulo, autor, paginas):
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas

    def descripcion(self):
        return f"'{self.titulo}' de {self.autor}, {self.paginas} páginas."

# Ejemplos
libro1 = Libro("Cien años de soledad", "Gabriel García Márquez", 417)
print(libro1.descripcion())

libro2 = Libro("El retrato de Dorian Grey", "Oscar Wilde.", 216)
print(libro2.descripcion())
```

```python
class Empleado:
    def __init__(self, nombre, salario):
        self.nombre = nombre
        self.salario = salario

    def aumento(self, porcentaje):
        incremento = self.salario * (porcentaje / 100)
        self.salario += incremento
        return f"Nuevo salario de {self.nombre}: ${self.salario:.2f}"

# Ejemplos
e1 = Empleado("Ana", 8500)
print(e1.aumento(10))

e2 = Empleado("Luis", 12000)
print(e2.aumento(5))
```
