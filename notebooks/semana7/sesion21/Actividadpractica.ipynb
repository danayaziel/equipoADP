```python
class Producto: 
    def __init__(self, nombre, precio, cantidad):
        self.nombre = nombre
        self.precio = precio
        self.cantidad = cantidad

    def valor_total(self):
        return self.precio * self.cantidad


# Lista de productos
productos = [
    Producto("Laptop", 15000, 2),
    Producto("Mouse", 300, 5),
    Producto("Teclado", 800, 3),
    Producto("Cargador", 2000, 1)
]

for p in productos:
    print(f"{p.nombre}: ${p.valor_total()}")
```
