#### 1. Ejemplos de lectura y escritura de archivos .txt.

```python
archivo = open("texto1.txt", "w")
archivo.write("Hola, este es el primer archivo de texto.\n")
archivo.write("Segunda línea de texto.\n")
archivo.close()

archivo = open("texto1.txt", "r")
print(archivo.read())
archivo.close()
```

```python
with open("texto2.txt", "w") as f:
    f.write("Hola, este es el segundo archivo de texto.\n")
    f.write("Segunda línea de texto.\n")

with open("texto2.txt", "r") as f:
    contenido = f.read()
    print("Contenido del archivo:")
    print(contenido)
```

```python
# Crear archivo de ejemplo
with open("frutas.txt", "w") as f:
    f.write("manzana\npera\naguacate\nplátano\nfresa\n")

# Leer con readlines()
with open("frutas.txt", "r") as f:
    lineas = f.readlines()
print("Contenido como lista:", lineas)

# Leer con ciclo for
with open("frutas.txt", "r") as f:
    for linea in f:
        print(linea.strip().upper())
```

#### 2. Ejemplos de lectura y escritura de archivos .csv

```python
import csv

with open("alumnos.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["nombre", "edad", "promedio"])
    writer.writerow(["Daniel", 23, 9.1])
    writer.writerow(["Jocelyn", 20, 8.5])
    writer.writerow(["Fatima", 22, 9.7])
```

```python
with open("alumnos.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    for fila in reader:
        print(fila)
```

```python
with open("alumnos.csv", "a", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Paola", 21, 8.9])
    writer.writerow(["Diego", 22, 9.2])
```

##### 4. Un ejercicio propio que combine ambos tipos de archivos (por ejemplo, leer datos de un .txt y guardarlos como .csv)

```python
with open("nombres.txt", "w") as f:
    f.write("Ana\n")
    f.write("Luis\n")
    f.write("María\n")
    f.write("Andrés\n")
    f.write("Paula\n")
```

```python
with open("nombres.txt", "r") as f:
    nombres = [linea.strip() for linea in f]
print(nombres)
```

```python
import csv

with open("nombres.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["nombre"])

    for nombre in nombres:
        writer.writerow([nombre])

print("Archivo nombres.csv creado correctamente.")
```

```python
nombres_A = [n for n in nombres if n.startswith("A")]

with open("nombres_A.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["nombre"])
    for n in nombres_A:
        writer.writerow([n])
```
