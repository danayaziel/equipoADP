### Actividad práctica

1) Leer el archivo

```python
with open("alumnos.txt", "r") as f:
    lineas = f.readlines()

print("Contenido en lista:", lineas)
```

2) Imprimir cada nombre en mayúsculas

```python
with open("alumnos.txt", "r") as f:
    for linea in f:
        print(linea.strip().upper())
```

3) Guardar los que empiezan con 'A' en un nuevo archivo 'alumnos_A.txt'

```python
nombres_A = []

with open("alumnos.txt", "r") as f:
    for linea in f:
        nombre = linea.strip()
        if nombre.startswith("A"): 
            nombres_A.append(nombre)

# Guardar en alumnos_A.txt
with open("alumnos_A.txt", "w") as f_out:
    for nombre in nombres_A:
        f_out.write(nombre + "\n")

print("Archivo 'alumnos_A.txt' creado con:", nombres_A)
```
