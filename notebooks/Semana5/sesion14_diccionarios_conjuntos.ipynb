## 🧩 Diccionarios (dict)
Un **diccionario** es una colección de pares **clave–valor**. Las claves son **únicas** y **hashables** (inmutables).


```python
# Creación
estudiante = {"nombre": "Ana", "edad": 21, "carrera": "Data Science", "aprobado": False, 1:3.1416}
print(estudiante)
```

```python

# Acceso por clave
print("Nombre:", estudiante["nombre"])  # puede lanzar KeyError si no existe
print("Edad (get):", estudiante.get("edad"))  # devuelve None si no existe
print("Ciudad (get con default):", estudiante.get("ciudad", "Desconocida"))
```

```python
print("Ciudad: ", estudiante["ciudad"])
```

```python
print("Posicion 0:", estudiante[0])
```

```python
print("Posicion 1:", estudiante[1])
```

```python
# Acceso por clave a un elemento inexistente
print("Fecha de nacimiento:", estudiante["fecha_nacimiento"])  #  lanza KeyError si no existe
```

```python
print(estudiante)
```

```python

# Modificación y agregado
estudiante["edad"] = 22
estudiante["semestre"] = 5
print("Modificado:", estudiante)
```

```python

# Eliminación
valor = estudiante.pop("semestre")  # elimina y devuelve el valor
print("Eliminado 'semestre':", valor)
```

```python
print(estudiante)
```

```python

# Métodos útiles
print("Claves:", list(estudiante.keys()))
print("Valores:", list(estudiante.values()))
print("Items:", list(estudiante.items()))
print(estudiante.items())
```

```python

# Recorrido recomendado
for clave, valor in estudiante.items():
    print(f"{clave} -> {valor}")
```

```python
for k in estudiante.keys():
    print(f"LLave: {k} -> Valor: {estudiante[k]}")
```


## 🧮 Conjuntos (set)
Un **conjunto** es una colección **no ordenada** de elementos **únicos**.
- Se usa para **eliminar duplicados** y para **operaciones de teoría de conjuntos**.


```python
# Creación
conjunto_a = {1, 2, 3, 3, 2}
conjunto_b = {3, 4, 5}
print("conjunto_a (sin duplicados):", conjunto_a)
print("conjunto_b:", conjunto_b)
```

```python

# Operaciones
print("Unión:", conjunto_a | conjunto_b)
print("Intersección:", conjunto_a & conjunto_b)
print("Diferencia a-b:", conjunto_a - conjunto_b)
print("Diferencia simétrica:", conjunto_a ^ conjunto_b)
```

```python

# Métodos
conjunto_a.add(10)
print("Agregando un elemento a conjunto_a:", conjunto_a)

```

```python

# Métodos
conjunto_a.add(1)
print("Agregando un elemento a conjunto_a:", conjunto_a)
```

```python

conjunto_a.discard(2)   
print("Eliminar un elemento existente de conjunto_a:", conjunto_a)
```

```python

conjunto_a.discard(2)   
print("Eliminar un elemento existente de conjunto_a:", conjunto_a)
```

```python

conjunto_a.discard(5)   # no lanza error si no existe
print("Eliminar un elemento NO existente en conjunto_a:", conjunto_a)
```


## 🔎 Diferencias clave
- **Diccionario**: pares `clave: valor`. Acceso por **clave**.
- **Conjunto**: colección de **valores únicos** sin pares.
- Ambos eliminan duplicados en sus **claves** (dict) o **elementos** (set), pero el dict guarda un valor asociado.


## 📌 Ejemplo en clase

```python
# Diccionario
estudiante = {"nombre": "Ana", "edad": 21, "carrera": "Data Science"}
print(estudiante["nombre"])

# Conjunto
conjunto_a = {1, 2, 3}
conjunto_b = {3, 4, 5}
print(conjunto_a & conjunto_b)  # intersección
```

### 1. Comprobar pertenencia con in en sets y dicts

Dada un conjunto de frutas y un diccionario de precios, pide al usuario una fruta y verifica si está disponible.

```python
frutas = {"manzana", "pera", "uva", "naranja"}
precios = {"manzana": 10, "pera": 12, "uva": 15}

f = input("Ingresa una fruta: ")

if f in frutas:
    print("La fruta está disponible.")
    if f in precios:
        print(f"Precio: {precios[f]} pesos")
    else:
        print("Precio no disponible.")
else:
    print("Esa fruta no está en el conjunto.")
```

```python
l = [1,2,3,4,5]
n = float(input("Ingresa un numero"))

if n in l:
    print(f"{n} existe en la lista")
else:
    print("f{n} NO disponible")
```

### 2. Crear un set para eliminar duplicados de una lista

Dada una lista de nombres con repeticiones, genera un conjunto de nombres únicos.

```python
nombres = ["Ana", "Luis", "Pedro", "Ana", "Marta", "Luis"]
nombres_unicos = set(nombres)

print("Lista original:", nombres)
print("Nombres únicos:", nombres_unicos)
```

### 3. Usar set.update(iterable) para unir varios elementos

Tienes un conjunto de números y quieres añadir una lista y una tupla de valores al mismo conjunto.

```python
numeros = {1, 2, 3}
lista_extra = [4, 5]
tupla_extra = (6, 7)

numeros.update(lista_extra)
numeros.update(tupla_extra)

print("Conjunto actualizado:", numeros)
```

### 4. Usar dict.setdefault(clave, valor_por_defecto)

Construir un diccionario que cuente cuántas veces aparece cada letra en una palabra, inicializando claves al vuelo.

```python
palabra = "programacion"
conteo = {}

for letra in palabra:
    conteo.setdefault(letra, 0)  # inicializa la clave si no existe
    conteo[letra] += 1

print("Conteo de letras:", conteo)
```

```python
"""
Ejercicio 1: Agenda de contactos
Usa un diccionario para agregar, buscar, actualizar y eliminar contactos.
"""

# Diccionario vacío
agenda = {}

# Agregar contactos
agenda["Ana"] = "555-1234"
agenda["Luis"] = "555-5678"
agenda["Marta"] = "555-8765"

# Buscar un contacto
nombre = "Luis"
if nombre in agenda:
    print(f"{nombre}: {agenda[nombre]}")
else:
    print(f"{nombre} no está en la agenda.")

# Actualizar número
agenda["Ana"] = "555-0000"

# Eliminar un contacto
agenda.pop("Marta", None)

# Mostrar agenda final
print("Agenda actual:", agenda)
```

```python
"""
Ejercicio 2: Frecuencia de palabras
Contar cuántas veces aparece cada palabra y mostrar las 3 más comunes.
"""

texto = "python es divertido y python es poderoso y python es útil"
palabras = texto.split()

# Crear diccionario de frecuencias
frecuencias = {}
for palabra in palabras:
    frecuencias[palabra] = frecuencias.get(palabra, 0) + 1

print("Frecuencia de palabras:", frecuencias)

# Obtener top 3 más comunes
top3 = sorted(frecuencias.items(), key=lambda x: x[1], reverse=True)[:3]
print("Top 3 palabras más comunes:", top3)
```

```python
"""
Ejercicio 3: Alumnos inscritos en dos materias
Usa conjuntos (set) para obtener intersección, unión y diferencia.
"""

matematicas = {"Ana", "Luis", "Pedro", "Sofía"}
fisica = {"Luis", "Sofía", "Carlos"}

interseccion = matematicas & fisica      # En ambas materias
union = matematicas | fisica             # En al menos una
solo_matematicas = matematicas - fisica  # Solo en matemáticas

print("En ambas:", interseccion)
print("En alguna:", union)
print("Solo matemáticas:", solo_matematicas)
```

```python
"""
Ejercicio 4: Nombres repetidos
Dado un listado con nombres repetidos, obtener los únicos y contarlos.
"""

nombres = ["Ana", "Luis", "Ana", "Pedro", "Luis", "Sofía", "Pedro"]
nombres_unicos = set(nombres)
print("Nombres únicos:", nombres_unicos)
print("Cantidad de nombres únicos:", len(nombres_unicos))
```

```python
"""
Ejercicio 5: Validar lista de correos
Detectar duplicados con set y marcar válidos/ inválidos con dict.
"""

correos = [
    "ana@gmail.com", "luis@hotmail.com", "marta@correo", 
    "luis@hotmail.com", "sofia@yahoo.com", "pedro.com"
]

# Detectar duplicados
duplicados = set([c for c in correos if correos.count(c) > 1])
print("Correos duplicados:", duplicados)

# Validar correos (deben tener '@' y '.')
validacion = {}
for correo in correos:
    if "@" in correo and "." in correo:
        validacion[correo] = "Válido"
    else:
        validacion[correo] = "Inválido"

print("Validación de correos:")
for k, v in validacion.items():
    print(f"{k}: {v}")
```
