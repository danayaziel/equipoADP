Mini‑proyecto: **Registro de estudiantes**  
Simula un registro con los campos: `nombre`, `edad`, `carrera`, `promedio`.

---


```

# Mini-proyecto: Registro de estudiantes
from typing import List, Dict, Any, Tuple

# Tipo para el registro
Registro = List[Dict[str, Any]]

# 🧍‍♀️ 1) Agregar estudiante
def agregar_estudiante(registro: Registro, estudiante: Dict[str, Any]) -> None:
    """
    Agrega un estudiante al registro validando los campos requeridos.
    Campos: nombre (str), edad (int), carrera (str), promedio (float 0-10)
    """
    assert isinstance(estudiante.get("nombre"), str) and estudiante["nombre"], "Nombre inválido"
    assert isinstance(estudiante.get("edad"), int) and estudiante["edad"] > 0, "Edad inválida"
    assert isinstance(estudiante.get("carrera"), str) and estudiante["carrera"], "Carrera inválida"
    assert isinstance(estudiante.get("promedio"), (int, float)) and 0 <= estudiante["promedio"] <= 10, "Promedio inválido"
    registro.append(estudiante)

# 📋 2) Listar estudiantes
def listar_estudiantes(registro: Registro) -> None:
    """Imprime una tabla con todos los estudiantes del registro."""
    if not registro:
        print("No hay estudiantes registrados.")
        return
    print(f"{'Nombre':<12} {'Edad':<4} {'Carrera':<12} {'Promedio':<8}")
    print("-" * 42)
    for est in registro:
        print(f"{est['nombre']:<12} {est['edad']:<4} {est['carrera']:<12} {est['promedio']:<8.2f}")

# 🎓 3) Buscar por carrera
def buscar_por_carrera(registro: Registro, carrera: str) -> Registro:
    """Devuelve una lista con los estudiantes de una carrera específica."""
    return [e for e in registro if e.get("carrera", "").lower() == carrera.lower()]

# 🏆 4) Mejor promedio
def mejor_promedio(registro: Registro) -> Tuple[str, float]:
    """Retorna el nombre y promedio del estudiante con mayor promedio."""
    if not registro:
        return ("", 0.0)
    mejor = max(registro, key=lambda e: e.get("promedio", 0))
    return (mejor["nombre"], mejor["promedio"])

# 💯 5) Extra: promedio general
def promedio_general(registro: Registro) -> float:
    """Calcula el promedio general de todos los estudiantes."""
    if not registro:
        return 0.0
    suma = sum(e["promedio"] for e in registro)
    return round(suma / len(registro), 2)
```

```

# ===========================
# 💻 EJEMPLO DE USO
# ===========================
registro: Registro = []

# Agregamos estudiantes de ejemplo
agregar_estudiante(registro, {"nombre": "Ana", "edad": 20, "carrera": "Datos", "promedio": 9.1})
agregar_estudiante(registro, {"nombre": "Luis", "edad": 22, "carrera": "IA", "promedio": 8.7})
agregar_estudiante(registro, {"nombre": "Marta", "edad": 21, "carrera": "Datos", "promedio": 9.5})
agregar_estudiante(registro, {"nombre": "Carlos", "edad": 23, "carrera": "Negocios", "promedio": 7.9})

print("📋 REGISTRO COMPLETO:")
listar_estudiantes(registro)

print("\n🎓 Estudiantes de la carrera 'Datos':")
listar_estudiantes(buscar_por_carrera(registro, "Datos"))

nombre, prom = mejor_promedio(registro)
print(f"\n🏆 Mejor promedio: {nombre} con {prom}")

print(f"\n💯 Promedio general del grupo: {promedio_general(registro)}")
```


## 🧠 Conclusiones
En este mini‑proyecto se usaron **listas** y **diccionarios**:
- La **lista** (`Registro`) permite almacenar múltiples estudiantes como si fuera una tabla.
- Cada **diccionario** representa un estudiante y facilita acceder a sus campos por **clave** (`nombre`, `edad`, `carrera`, `promedio`).
- Las operaciones de **búsqueda** y **agregación** (máximo y promedio) son simples con comprensiones, `max`, y `sum`.

Esta combinación es adecuada para prototipos de “mini‑bases de datos” en memoria por su **claridad** y **flexibilidad**.
