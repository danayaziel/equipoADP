# Ejercicios de Bucles
Estos ejercicios utilizan `for` y `while` para resolver diferentes problemas sin usar `input()`.

## 1. Sumatoria de los primeros _n_ números

```python
n = 15
suma = 0
for i in range(1, n + 1):
    suma += i
print(f"Suma de los primeros {n} números:", suma)
```

## 2. Cálculo del factorial de un número

```python
numero = 6
factorial = 1
for i in range(1, numero + 1):
    factorial *= i
print(f"El factorial de {numero} es {factorial}")
```

## 3. Contador de vocales en una palabra

```python
palabra = "inteligencia"
vocales = "aeiou"
total_vocales = 0
for letra in palabra:
    if letra in vocales:
        total_vocales += 1
print(f"La palabra '{palabra}' tiene {total_vocales} vocales")
```

## 4. Lista de cuadrados del 1 al 10

```python
cuadrados = []
for i in range(1, 11):
    cuadrados.append(i ** 2)
print("Cuadrados del 1 al 10:", cuadrados)
```

## 5. Simulación de cajero con intentos limitados

```python
pin_correcto = "2468"
intentos = 3
pins_ingresados = ["1234", "0000", "2468"]
for intento in pins_ingresados:
    if intento == pin_correcto:
        print("Acceso permitido con PIN:", intento)
        break
    else:
        intentos -= 1
        print(f"PIN incorrecto ({intento}). Intentos restantes: {intentos}")
if intentos == 0:
    print("Tarjeta bloqueada")
```
