## 1. Tabla de multiplicar de un número fijo

```python
numero = 7
print(f"Tabla de multiplicar del {numero}:")
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

## 2. Simulación de contraseña correcta con while

```python
contraseña_correcta = "python123"
intentados = ["abc", "12345", "python123"]
indice = 0
while indice < len(intentados) and intentados[indice] != contraseña_correcta:
    print(f"Intento fallido: {intentados[indice]}")
    indice += 1
if indice < len(intentados):
    print("¡Acceso concedido!")
else:
    print("Contraseña no encontrada en los intentos")
```

## 3. Uso de break y continue en un bucle

```python
for i in range(1, 10):
    if i == 5:
        print("Se encontró el 5, se interrumpe el bucle con break")
        break
    if i % 2 == 0:
        continue
    print("Número impar:", i)
else:
    print("Fin del bucle")
```
