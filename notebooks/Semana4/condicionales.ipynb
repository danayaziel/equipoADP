Tres ejercicios usando if, elif y else.

```python
# 1. Verificar si una persona puede entrar al cine
edad = 15
if edad < 12:
    print("Entrada gratis para menores de 12")
elif edad <= 17:
    print("Entrada con descuento para adolescentes")
else:
    print("Entrada normal para adultos")
```

```python
# 2. Clasificar un número como negativo, cero o positivo
numero = -3
if numero < 0:
    print("El número es negativo")
elif numero == 0:
    print("El número es cero")
else:
    print("El número es positivo")
```

```python
# 3. Determinar el tipo de triángulo según las longitudes
lado1 = 5
lado2 = 5
lado3 = 8
if lado1 == lado2 == lado3:
    print("Triángulo equilátero")
elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
    print("Triángulo isósceles")
else:
    print("Triángulo escaleno")
```
