# Actividad práctica: Condicionales

```python
# 1. Programa que solicita una calificación y muestra si está aprobado o reprobado
calificacion = int(input("Ingresa tu calificación: "))
if calificacion >= 60:
    print("Aprobado")
else:
    print("Reprobado")
```

```python
# 2. Programa que solicita la temperatura y muestra un mensaje
temperatura = float(input("Ingresa la temperatura en °C: "))
if temperatura < 15:
    print("Frío")
elif temperatura <= 25:
    print("Templado")
else:
    print("Caluroso")
```

```python
# 3. Uso de operador ternario para par o impar
numero = int(input("Ingresa un número: "))
resultado = "Par" if numero % 2 == 0 else "Impar"
print("El número es:", resultado)
```
