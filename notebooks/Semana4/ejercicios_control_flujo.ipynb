## 1. Validador sencillo de contraseña
La idea es revisar si una contraseña cumple con ciertas reglas básicas sin pedir input.

```python
contraseña = "Python1234"
longitud_ok = len(contraseña) >= 8
contiene_digito = any(c.isdigit() for c in contraseña)
contiene_letra = any(c.isalpha() for c in contraseña)

if longitud_ok and contiene_digito and contiene_letra:
    print("Contraseña válida")
else:
    print("Contraseña no cumple con los requisitos")
```

## 2. Secuencia de Fibonacci hasta n y conteo de pares/impares

```python
n = 10
fibonacci = [0, 1]
for i in range(2, n):
    fibonacci.append(fibonacci[-1] + fibonacci[-2])

pares = 0
impares = 0
for num in fibonacci:
    if num % 2 == 0:
        pares += 1
    else:
        impares += 1

print("Secuencia Fibonacci:", fibonacci)
print("Pares:", pares, "Impares:", impares)
```

## 3. Piedra, papel o tijera vs computadora (marcador al mejor de 3)

```python
import random

opciones = ["piedra", "papel", "tijera"]
puntuacion_jugador = 0
puntuacion_cpu = 0

# Simular 5 rondas fijas, sin input, para no hacerlo repetitivo
rondas = ["piedra", "papel", "tijera", "piedra", "papel"]

for eleccion_jugador in rondas:
    cpu = random.choice(opciones)

    if eleccion_jugador == cpu:
        pass
    elif (
        (eleccion_jugador == "piedra" and cpu == "tijera") or
        (eleccion_jugador == "papel" and cpu == "piedra") or
        (eleccion_jugador == "tijera" and cpu == "papel")
    ):
        puntuacion_jugador += 1
    else:
        puntuacion_cpu += 1

# Mostrar resultado final
print("Puntuación final -> Jugador:", puntuacion_jugador, "CPU:", puntuacion_cpu)
if puntuacion_jugador > puntuacion_cpu:
    print("Ganaste")
elif puntuacion_cpu > puntuacion_jugador:
    print("Perdiste")
else:
    print("Empate")
```
