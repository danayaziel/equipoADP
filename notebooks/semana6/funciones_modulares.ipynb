### Funciones para limpiar datos

Se utilizó el dataset Iris.csv de Kaggle: https://www.kaggle.com/datasets/uciml/iris 

```python
import csv
from libreria1.limpieza import eliminar_duplicados, reemplazar_nulos, convertir_tipos

# 1) Cargar CSV a lista de dicts
with open("Iris.csv", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    data = list(reader)

print("Filas originales:", len(data))

# 2) Eliminar duplicados por 'id'
for i, fila in enumerate(data):
    fila["id"] = i
sin_dups = eliminar_duplicados(data, clave="id")
print("Tras eliminar duplicados:", len(sin_dups))

# 3) Reemplazar nulos/vacíos por valores por defecto
sustitutos = {"SepalLengthCm": "0", "SepalWidthCm": "0", "PetalLengthCm": "0", "PetalWidthCm": "0", "Species": "Unknown"}
rellenos = reemplazar_nulos(sin_dups, sustitutos)

# 4) Convertir tipos
def to_int_safe(x):
    try:
        return int(x)
    except Exception:
        return None

def to_float_safe(x):
    try:
        return float(x)
    except Exception:
        return None

convertidos = convertir_tipos(rellenos, {
    "SepalLengthCm": to_float_safe,
    "SepalWidthCm": to_float_safe,
    "PetalLengthCm": to_float_safe,
    "PetalWidthCm": to_float_safe
})

# Vista previa (primeras 5 filas)
for fila in convertidos[:5]:
    print(fila)
```

### Estadísticas

```python
from libreria.estadistica_basica import media, mediana, moda, rango

sepal_lengths = [f["SepalLengthCm"] for f in convertidos if f["SepalLengthCm"] is not None]
sepal_widths  = [f["SepalWidthCm"] for f in convertidos if f["SepalWidthCm"] is not None]
petal_lengths = [f["PetalLengthCm"] for f in convertidos if f["PetalLengthCm"] is not None]
petal_widths  = [f["PetalWidthCm"] for f in convertidos if f["PetalWidthCm"] is not None]

print("SEPAL LENGTH")
print("Media:", round(media(sepal_lengths),2))
print("Mediana:", mediana(sepal_lengths))
print("Moda:", moda(sepal_lengths))
print("Rango:", round(max(sepal_lengths) - min(sepal_lengths),2))

print("\nSEPAL WIDTH")
print("Media:", round(media(sepal_widths),2))
print("Mediana:", mediana(sepal_widths))
print("Moda:", moda(sepal_widths))
print("Rango:", round(max(sepal_widths) - min(sepal_widths),2))

print("\nPETAL LENGTH")
print("Media:", round(media(petal_lengths),2))
print("Mediana:", mediana(petal_lengths))
print("Moda:", moda(petal_lengths))
print("Rango:", max(petal_lengths) - min(petal_lengths))

print("\nETAL WIDTH")
print("Media:", round(media(petal_widths),2))
print("Mediana:", mediana(petal_widths))
print("Moda:", moda(petal_widths))
print("Rango:", max(petal_widths) - min(petal_widths))
```
