### 1. Función matemática

```python
def factorial(n):
    """
    Calcula el factorial de un número entero positivo n.
    Ejemplo: 
    El programa muestra: Introduce un número entero positivo
    entrada: 6
    salida: El factorial de 6 es: 720
    """
    if n < 0: 
        print("No se puede calcular el factorial, introduce un número positivo")
        return None
    elif n == 0 or n == 1:
        return 1
    else: 
        factorial = 1
        contador = 1
        while contador <= n:
            factorial = factorial * contador
            contador = contador + 1
        return factorial

n = int(input("Introduce un número entero positivo: "))
print(f"El factorial de {n} es: {factorial(n)}")

# Prueba
print(f"El factorial de 5 es: {factorial(5)}")
```

### 2. Función con cadenas

```python
def contar_vocales(texto):
    """Cuenta el número de vocales en 'texto' (a, e, i, o, u). Si hay una vocal en mayúscula la convierte en minúscula 
    Ejemplo: 
    El programa muestra: Introduce tu palabra o texto
    entrada: Estados Unidos Mexicanos
    salida: 'Estados Unidos Mexicanos' tiene 10 vocales
    """
    texto = texto.lower()
    vocales = "aeiou"
    contador = 0
    for letra in texto:
        if letra in vocales:
            contador += 1
    return contador
    
texto = input("Introduce tu palabra o texto: ")
print(f"'{texto}' tiene {contar_vocales(texto)} vocales")

# Prueba
print("Vocales en 'Bienvenidos a clase':", contar_vocales("Bienvenidos a clase"))
```

### 3. Función con listas

```python
def maximo_lista(numeros):
    """Verifica cual es el número máximo en una lista de números. Devuelve el máximo de la lista sin usar max(); None si está vacía.
    Ejemplo: 
    El programa imprime: Introduce los valores
    entrada: 5 10 13 2 8 3
    salida: El número máximo de [5 10 13 2 8 3] es 13.0
    """
    if len(numeros) == 0:
        return None
    else:
        maximo = numeros[0]
        for num in numeros:
            if num > maximo:
                maximo = num
        return maximo

lista = input("Introduce los valores: ")
numeros = [float(i) for i in lista.split()]
print(f"El número máximo de [{lista}] es {maximo_lista(numeros)}")

# Prueba
print("Máximo de [3, 9, 2, 10]:", maximo_lista([3, 9, 2, 10]))
```

### 4. Función lógica

```python
def es_primo_v2(n):
    """ Verifica si un número es primo. Devuelve True si n es un entero primo; de lo contrario False.
    Ejemplo: 
    El programa muestra: Introduce un número
    entrada: 27
    salida: True
    """
    if n % 2 == 0:
        return False
    else:
        return True
    
n = int(input("Introduce un número: "))
print(es_primo_v2(n))

# Prueba
print("¿11 es primo?:", es_primo_v2(11))
```

### 5. Función propia (Anagrama)

```python
def es_anagrama(texto1, texto2):
    """Determina si dos frases son anagramas, ignorando mayúsculas, espacios y signos de puntuación. 
    Ejemplo: 
    El programa muestra: 
    Introduce la primera frase 
    entrada: William Shakespeare
    Introduce la segunda frase
    entrada: I am a weakish speller
    salida: 'William Shakespeare' y 'I am a weakish speller' son anagramas
    """
    texto1_limpio = texto1.replace(" ", "").lower()
    texto2_limpio = texto2.replace(" ", "").lower()

    return sorted(texto1_limpio) == sorted(texto2_limpio)
    
texto1 = input("Introduce la primera frase: ")
texto2 = input("\nIntroduce la segunda frase: ")
if es_anagrama(texto1, texto2):
    print(f"'{texto1}' y '{texto2}' son anagramas.")
else:
    print(f"'{texto1}' y '{texto2}' no son anagramas.")

# Prueba
print("¿'Japones' y 'Esponja' son o no anagramas?", es_anagrama("Japones","Esponja"))
```
