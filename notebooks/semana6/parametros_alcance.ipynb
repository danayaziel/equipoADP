### 1. Potencia (base, exponente=2)

```python
def potencia(base, exponente=2):
    """
    Calcula la potencia de un número.

    Parámetros:
     base (float): número base.
     exponente (int, opcional): exponente al cual elevar la base. Por defecto es 2.

    Retorna:
     float: resultado de base elevado al exponente.
    """
    return base ** exponente


# Ejemplos de uso
print("Ejemplo 1: potencia(5) =", potencia(5))        # 25
print("Ejemplo 2: potencia(2, 4) =", potencia(2, 4))  # 16

# Pruebas
print("Prueba: 3^2 =", potencia(3))                   # 9
print("Prueba: 2^3 =", potencia(2, 3))                # 8
```

### 2. Precio_con_iva (precio, iva=0.16)

```python
def precio_con_iva(precio, iva=0.16):
    """
    Calcula el precio final con IVA incluido.

    Parámetros:
     precio (float): precio original del producto.
     iva (float, opcional): tasa de IVA. Por defecto es 0.16 (16%).

    Retorna:
     float: precio final con IVA.

    Ejemplo:
    precio_con_iva(100)
    116.0
    precio_con_iva(200, 0.08)
    216.0
    """
    total = precio * (1 + iva)
    return round(total, 2)

# Ejemplos de uso
print(precio_con_iva(100))        # 116.0
print(precio_con_iva(200, 0.08))  # 216.0

# Pruebas
assert precio_con_iva(100) == 116.0
assert precio_con_iva(200, 0.1) == 220.0
```

### 4. Descuento (precio, porcentaje=0.10)

```python
def descuento(precio, porcentaje=0.10):
    """
    Calcula el precio con descuento aplicado.

    Parámetros:
     precio (float): precio original.
     porcentaje (float, opcional): porcentaje de descuento (por defecto 0.10, en porcentaje 10%).

    Retorna:
     float: precio final con el descuento aplicado.
    """
    return precio * (1 - porcentaje)


# Ejemplos de uso
print("Ejemplo 1: descuento(100) =", descuento(100))          # 90.0
print("Ejemplo 2: descuento(250, 0.2) =", descuento(250, 0.2)) # 200.0

# Pruebas
print("Prueba: descuento del 10% a 100 =", descuento(100))     # 90.0
print("Prueba: descuento del 25% a 200 =", descuento(200, 0.25)) # 150.0
```

### ¿Por qué evitar mutables como valores por defecto?

Se deben evitar los valores mutables como argumentos por defecto en Python porque el valor por defecto 
(como una lista o diccionario) se crea una sola vez cuando se define la función, y ese mismo objeto se 
reutiliza en todas las llamadas posteriores a la función. 
Esto significa que cualquier modificación hecha al valor por defecto en una llamada a la función se 
conservará para las llamadas siguientes, provocando un comportamiento inesperado y difícil de depurar.
