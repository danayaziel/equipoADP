```python
import pandas as pd
#creamos el data frame base
empleados = pd.DataFrame({
    "Nombre": ["Ana", "Luis", "Carlos", "María", "Jorge"],
    "Edad": [22, 35, 29, 41, 19],
    "Salario_mensual": [9000, 15000, 12000, 18000, 8000],
    "Genero": ["F", "M", "M", "F", "M"]
})

empleados
```

```python
#creamos nueva columna (salario anual)

empleados["Salario_anual"] = empleados["Salario_mensual"] * 12
empleados
```

```python
#creamos una columna donde categorice por edades

empleados["Categoria_edad"] = empleados["Edad"].apply(
    lambda x: "Joven" if x < 30 else "Adulto"
)

empleados
```

```python
empleados["Genero_completo"] = empleados["Genero"].replace({
    "F": "Femenino",
    "M": "Masculino"
})

empleados
```
