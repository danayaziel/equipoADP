```python
import pandas as pd
#Primero creamos el data frame para los ejemplos

empleados = pd.DataFrame({
    "Nombre": ["Ana", "Luis", "Marcos", "Sofía", "Daniel", "Karla", "Jorge", "Elena"],
    "Edad":   [28, 35, 42, 30, 25, 33, 45, 29],
    "Salario": [12000, 18000, 9500, 22000, 8000, 16000, 30000, 11000],
    "Departamento": ["Ventas", "TI", "Recursos Humanos", "Marketing", "TI", "Ventas", "Dirección", "Marketing"]
})

empleados
```

```python
#ejemplo 1 Filtrar empleados con salario mayor a $10,000

empleados[empleados["Salario"] > 10000]
```

```python
#ejemplo 2 Filtrar empleados con edad mayor a 30 y salario mayor a $15,000

empleados[(empleados["Edad"] > 30) & (empleados["Salario"] > 15000)]
```

```python
#ejemplo 3 Filtrar empleados que NO sean del departamento de TI

empleados[empleados["Departamento"] != "TI"]
```

```python
#ejemplo 4 Empleados entre 25 y 35 años

empleados[(empleados["Edad"] >= 25) & (empleados["Edad"] <= 35)]
```

```python
#ejemplo 5 Filtrar empleados cuyo nombre empieza con la letra “A” o “E”

empleados[empleados["Nombre"].str.startswith(("A", "E"))]
```
