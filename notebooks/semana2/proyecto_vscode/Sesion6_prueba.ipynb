```python
import random
import numpy as np
import matplotlib.pyplot as plt

# Lista de 100 números aleatorios
numeros = [random.randint(1, 100) for _ in range(100)]

# Media y varianza
media = np.mean(numeros)
varianza = np.var(numeros)
print("Media:", media)
print("Varianza:", varianza)

# Histograma
plt.hist(numeros, bins=10)
plt.title("Histograma de números aleatorios")
plt.xlabel("Valores")
plt.ylabel("Frecuencia")
plt.show()
```
