Ejemplos de conversiones en Python con comentarios explicativos.

```python
# Ejemplo 1: Convertir string a entero
texto_num = "50"  # Creamos un string que contiene un número
numero_entero = int(texto_num)  # Convertimos el string a un entero
print(numero_entero, type(numero_entero))
```

```python
# Ejemplo 2: Convertir string a flotante
texto_decimal = "3.1416"  # String con un valor decimal
numero_float = float(texto_decimal)  # Convertimos el string a float
print(numero_float, type(numero_float))
```

```python
# Ejemplo 3: Convertir entero a string
entero_valor = 2025  # Número entero
texto_convertido = str(entero_valor)  # Lo convertimos a string
print(texto_convertido, type(texto_convertido))
```
