# Actividad práctica: Operadores y expresiones

```python
# 1. Calcular el área de un círculo
pi = 3.1416
radio = 5
area = pi * (radio ** 2)
print("Área del círculo de radio 5:", area)
```

```python
# 2. Verificar si un número es par o impar
numero = 7
if numero % 2 == 0:
    print(numero, "es par")
else:
    print(numero, "es impar")
```

```python
# 3. Verificar si una persona puede votar sin input
edad = 16
if edad >= 18:
    print("Puedes votar")
else:
    print("No puedes votar aún")
```
