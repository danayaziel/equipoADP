# Cálculo del Índice de Masa Corporal (IMC)
Este programa calcula el **Índice de Masa Corporal (IMC)** a partir del peso y la altura proporcionados por el usuario.

La fórmula utilizada es:

$$ IMC = \frac{peso}{altura^2} $$

**Peso:** se ingresa en kilogramos (kg).
<br>**Altura:** se ingresa en metros (m).

El resultado permite evaluar el estado corporal según rangos establecidos.

```python
# Cálculo del IMC usando input y siguiendo buenas prácticas (PEP8)

# Solicitar los datos al usuario
peso = float(input("Ingresa tu peso en kg: "))  # Peso en kilogramos
altura = float(input("Ingresa tu altura en metros: "))  # Altura en metros

# Cálculo del IMC
imc = peso / (altura ** 2)

# Mostrar el resultado
print("\nTu IMC es:", round(imc, 2))

# Clasificación según el resultado
if imc < 18.5:
    print("Clasificación: Bajo peso")
elif 18.5 <= imc < 25:
    print("Clasificación: Peso normal")
elif 25 <= imc < 30:
    print("Clasificación: Sobrepeso")
else:
    print("Clasificación: Obesidad")
```
