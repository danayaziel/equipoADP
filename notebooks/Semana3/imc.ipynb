
# 🧮 Cálculo del IMC (Índice de Masa Corporal)

Este cuaderno calcula el **IMC** a partir del **peso (kg)** y la **altura (m)** siguiendo buenas prácticas de PEP 8.  
Incluye:
- Entrada de datos con `input()`.
- Funciones reutilizables para **cálculo** y **clasificación**.
- **Pruebas** con distintos valores de peso y altura.
- Comentarios explicando cada paso.


```

# -*- coding: utf-8 -*-
"""
Módulo de utilidad para calcular y clasificar el IMC.
- Fórmula del IMC: IMC = peso / (altura ** 2)
- Peso en kilogramos (kg)
- Altura en metros (m)
"""

def calcular_imc(peso: float, altura: float) -> float:
    """Calcula el IMC con la fórmula peso / (altura ** 2).
    
    Args:
        peso (float): Peso en kilogramos. Debe ser > 0.
        altura (float): Altura en metros. Debe ser > 0.
        
    Returns:
        float: Valor de IMC (no redondeado).
        
    Raises:
        ValueError: Si peso o altura no son mayores que cero.
    """
    if peso <= 0 or altura <= 0:
        raise ValueError("El peso y la altura deben ser mayores que cero.")
    return peso / (altura ** 2)


def clasificar_imc(imc: float) -> str:
    """Clasifica el IMC de acuerdo con rangos estándar.
    
    Rango de referencia:
      - IMC < 18.5           -> "Bajo peso"
      - 18.5 <= IMC < 25     -> "Peso normal"
      - 25 <= IMC < 30       -> "Sobrepeso"
      - IMC >= 30            -> "Obesidad"
    
    Args:
        imc (float): Índice de Masa Corporal.
        
    Returns:
        str: Etiqueta de clasificación.
    """
    if imc < 18.5:
        return "Bajo peso"
    if imc < 25:
        return "Peso normal"
    if imc < 30:
        return "Sobrepeso"
    return "Obesidad"
```

```

# ===============================================================
# 1) Entrada de datos por teclado usando input()
#    - Se convierten a float
#    - Se valida que sean positivos
#    - Se calcula y muestra el IMC y su clasificación
# ===============================================================
try:
    # Solicitar al usuario su peso y altura
    peso_str = input("Ingresa tu peso en kg (ej. 68): ").strip()
    altura_str = input("Ingresa tu altura en metros (ej. 1.72): ").strip()
    
    # Convertir entradas a números reales (float)
    peso = float(peso_str)
    altura = float(altura_str)
    
    # Validar que ambos sean mayores que cero
    if peso <= 0 or altura <= 0:
        raise ValueError("El peso y la altura deben ser números positivos mayores que cero.")
    
    # Calcular el IMC y su clasificación
    imc_valor = calcular_imc(peso, altura)
    etiqueta = clasificar_imc(imc_valor)
    
    # Mostrar resultados redondeando el IMC a 2 decimales
    print(f"Tu IMC es: {imc_valor:.2f}")
    print(f"Clasificación: {etiqueta}")
    
except ValueError as exc:
    # Manejo de errores: entradas no numéricas o fuera de rango
    print(f"Entrada no válida: {exc}")
```

```

# ===============================================================
# 2) Pruebas con distintos valores de peso y altura
#    - Útil para verificar rápidamente la lógica del programa
#    - No requiere interacción del usuario
# ===============================================================

# Lista de casos de prueba (peso en kg, altura en m)
casos_prueba = [
    {"peso": 50, "altura": 1.70},   # IMC bajo/normal dependiendo del valor exacto
    {"peso": 68, "altura": 1.72},   # Caso de ejemplo del enunciado
    {"peso": 80, "altura": 1.70},   # Cerca del rango de sobrepeso
    {"peso": 95, "altura": 1.62},   # Probable obesidad
    {"peso": 120, "altura": 1.80},  # Obesidad
    {"peso": 60, "altura": 1.80},   # Bajo/normal
]

# Calcular resultados para cada caso
resultados = []
for caso in casos_prueba:
    imc_val = calcular_imc(caso["peso"], caso["altura"])
    resultados.append({
        "peso_kg": caso["peso"],
        "altura_m": caso["altura"],
        "imc": round(imc_val, 2),
        "clasificación": clasificar_imc(imc_val),
    })

# Mostrar tabla de resultados
try:
    import pandas as pd
    df = pd.DataFrame(resultados)
    display(df)
except Exception:
    # Si no se puede usar pandas, imprimir en formato simple
    print("peso_kg | altura_m | imc  | clasificación")
    for r in resultados:
        print(f"{r['peso_kg']:7} | {r['altura_m']:.2f}   | {r['imc']:.2f} | {r['clasificación']}")
```
