# Cálculo del Índice de Masa Corporal (IMC)
Este notebook calcula el **Índice de Masa Corporal (IMC)** usando la fórmula:

$$ IMC = \frac{peso}{altura^2} $$

El peso se mide en kilogramos y la altura en metros. El resultado permite clasificar el estado de salud con base en rangos establecidos.

```python
# Cálculo del IMC con valores de ejemplo
# Peso y altura definidos directamente (hardcodeados)
peso = 70       # peso en kilogramos
altura = 1.75   # altura en metros

# Fórmula del IMC
imc = peso / (altura ** 2)

# Salida del resultado y clasificación
print("Tu IMC es:", imc)

if imc < 18.5:
    print("Bajo peso")
elif 18.5 <= imc < 25:
    print("Peso normal")
elif 25 <= imc < 30:
    print("Sobrepeso")
else:
    print("Obesidad")
```

## Ejemplo adicional con otros valores
