## Problema 1: Área de un triángulo

```

# Datos de entrada
base = 10     # base del triángulo en unidades (p. ej., cm)
altura = 6    # altura del triángulo en las mismas unidades

# Cálculo
# Fórmula del área del triángulo: (base * altura) / 2
area_triangulo = (base * altura) / 2

# Salida
print("Área del triángulo:", area_triangulo)
```

## Problema 2: Verificar múltiplo

```

# Número a evaluar
numero = 24  # probaremos si 24 es múltiplo de 6

# Operador aritmético % (módulo) devuelve el residuo de la división
# Si el residuo es 0, entonces 'numero' es múltiplo de 6
if numero % 6 == 0:   # Operador relacional == compara igualdad
    print(numero, "es múltiplo de 6")
else:
    print(numero, "NO es múltiplo de 6")
```

## Problema 3: Mayor de edad con INE

```

# Datos de la persona
edad = 20        # años cumplidos
tiene_INE = True # booleano (True/False)

# Condición compuesta con operadores relacionales y lógicos
# edad >= 18  -> verifica si es mayor o igual a 18
# and         -> exige que ambas condiciones sean verdaderas
# tiene_INE   -> debe ser True
if edad >= 18 and tiene_INE:
    print("La persona puede votar")
else:
    print("La persona NO puede votar")
```

## Problema 4: Número en rango [10, 20] (inclusive)

```

# Valor a revisar
num = 15

# Forma directa con operadores relacionales y lógicos
if num >= 10 and num <= 20:
    print(num, "está en el rango 10-20")
else:
    print(num, "NO está en el rango 10-20")

# Forma alternativa (equivalente y más "pitónica") usando comparación encadenada:
# if 10 <= num <= 20:
#     print(num, "está en el rango 10-20")
# else:
#     print(num, "NO está en el rango 10-20")
```

## Problema 5: Expresión compuesta

```

# Resolvemos paso a paso la expresión:
# (5 + 3) * 2 > 10 and not (4 == 5)
# 1) (5 + 3) * 2 -> 8 * 2 = 16
# 2) 16 > 10     -> True
# 3) (4 == 5)    -> False
# 4) not False   -> True
# 5) True and True -> True
resultado = (5 + 3) * 2 > 10 and not (4 == 5)
print("Resultado de la expresión:", resultado)
```

- **Relacionales**: `==`, `>=`, `<=`, `>`
- **Lógicos**: `and`, `not`
