# Actividad práctica: Tipos de datos primitivos y conversiones

```python
# Variables de cada tipo
mi_entero = 42
mi_decimal = 9.81
mi_logico = False
mi_texto = "Python es genial"

print("Ejemplos:")
print(mi_entero, type(mi_entero))
print(mi_decimal, type(mi_decimal))
print(mi_logico, type(mi_logico))
print(mi_texto, type(mi_texto))
```

```python
# Conversiones
ejemplo1 = int("100")
ejemplo2 = float("45.67")
ejemplo3 = str(2025)

print("\nConversiones:")
print(ejemplo1, type(ejemplo1))
print(ejemplo2, type(ejemplo2))
print(ejemplo3, type(ejemplo3))
```
